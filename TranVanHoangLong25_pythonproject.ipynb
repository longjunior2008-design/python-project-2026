```python
#1.Tạo frame
data = {
    'Họ và tên': ['An', 'Bình', 'Chi', 'Dung', 'Hà', 'Lan', 'Minh', 'Nam', 'Phúc', 'Trang'],
    'Chuyên cần': [8.5, 9.0, 7.5, 8.0, 9.0, 6.5, 8.0, 7.0, 8.5, 9.5],
    'Giữa kỳ': [7.0, 8.0, 6.5, 7.5, 8.5, 5.5, 7.0, 6.0, 7.5, 8.0],
    'Cuối kỳ': [8.0, 8.5, 7.0, 8.0, 9.0, 6.0, 7.5, 6.5, 8.0, 9.0]
}

df = pd.DataFrame(data)

print("YÊU CẦU 1: BẢNG DỮ LIỆU BAN ĐẦU")
display(df)
print("\n" + "-"*50 + "\n")

#2.Tính điểm tổng kết
# Công thức: Điểm tổng kết = 20% * Chuyên cần + 30% * Giữa kỳ + 50% * Cuối kỳ
df['Tổng kết'] = (df['Chuyên cần'] * 0.2) + (df['Giữa kỳ'] * 0.3) + (df['Cuối kỳ'] * 0.5)
df['Tổng kết'] = df['Tổng kết'].round(2) # Làm tròn 2 chữ số thập phân

print("YÊU CẦU 2: BẢNG DỮ LIỆU CÓ THÊM ĐIỂM TỔNG KẾT")
display(df[['Họ và tên', 'Chuyên cần', 'Giữa kỳ', 'Cuối kỳ', 'Tổng kết']])
print("\n" + "-"*50 + "\n")

#3.Xếp loại
def xep_loai(diem):
    if diem >= 8.5:
        return 'Giỏi'
    elif diem >= 7.0:
        return 'Khá'
    elif diem >= 5.0:
        return 'Trung bình'
    else:
        return 'Yếu'

df['Xếp loại'] = df['Tổng kết'].apply(xep_loai)

print("YÊU CẦU 3: BẢNG DỮ LIỆU SAU KHI XẾP LOẠI")
display(df)
print("\n" + "-"*50 + "\n")

#4.Thống kê
print("YÊU CẦU 4: THỐNG KÊ KẾT QUẢ HỌC TẬP")

# 1. Điểm tổng kết trung bình của lớp
dtb_lop = df['Tổng kết'].mean()
print(f"• Điểm tổng kết trung bình của lớp: {dtb_lop:.2f}")

# 2. Sinh viên có điểm tổng kết cao nhất
max_sv = df.loc[df['Tổng kết'].idxmax()]
print(f"• Sinh viên có điểm tổng kết cao nhất: {max_sv['Họ và tên']} (Điểm: {max_sv['Tổng kết']})")

# 3. Sinh viên có điểm tổng kết thấp nhất
min_sv = df.loc[df['Tổng kết'].idxmin()]
print(f"• Sinh viên có điểm tổng kết thấp nhất: {min_sv['Họ và tên']} (Điểm: {min_sv['Tổng kết']})")

# 4. Đếm số sinh viên đạt (điểm tổng kết >= 5.0)
so_sv_dat = (df['Tổng kết'] >= 5.0).sum()
print(f"• Số sinh viên đạt (điểm tổng kết >= 5.0): {so_sv_dat} / {len(df)}")
print("\n" + "-"*50 + "\n")

#5.Trực quan hóa bằng Matplotlib
print("YÊU CẦU 5: TRỰC QUAN HÓA")

# Lấy tên ngắn (từ cuối cùng của họ và tên) để biểu đồ gọn gàng giống ảnh mẫu
short_names = df['Họ và tên'].apply(lambda x: x.split()[-1])
scores = df['Tổng kết']

plt.figure(figsize=(9, 5))

# Vẽ biểu đồ cột ngang (barh) với màu xanh dương đặc trưng
plt.barh(short_names, scores, color='#5B92E5')

# Thêm tiêu đề và tên các trục theo yêu cầu[span_32](start_span)[span_32](end_span)[span_33](start_span)[span_33](end_span)
plt.xlabel('Điểm tổng kết', fontsize=11)
plt.ylabel('Sinh viên', fontsize=11)
plt.title(
    'Biểu đồ cột thể hiện điểm tổng kết của 10 sinh viên',
    fontsize=12,
    fontweight='bold',
)

# Đảo ngược trục y để sinh viên đầu danh sách hiển thị ở trên cùng
plt.gca().invert_yaxis()

# Thêm lưới mờ dọc trục x để quan sát điểm số trực quan hơn
plt.grid(axis='x', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
